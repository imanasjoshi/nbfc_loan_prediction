# Feature Engineering

Creating new features from the raw data.

```
import pandas as pd
import numpy as np
train = pd.read_csv('../data/raw/Train.csv')
```

```
train['credit_utilization'] = np.where(train['total_revolving_limit'] > 0, train['revolving_balance'] / train['total_revolving_limit'], 0)
```

```
train['loan_to_income'] = np.where(train['annual_income'] > 0, train['loan_amnt'] / train['annual_income'], 0)
```
