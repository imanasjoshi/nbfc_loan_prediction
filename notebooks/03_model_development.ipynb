# Model Development

Training baseline and advanced models.

```
import sys
sys.path.append('..')
from src.data_preprocessing import load_data, clean_data
from src.feature_engineering import create_features
import src.config as config
from catboost import CatBoostClassifier
from sklearn.model_selection import train_test_split
```

```
train, test = load_data()
train_c = clean_data(train)
train_f = create_features(train_c)
```

```
X = train_f[config.ORIGINATION_FEATURES]
y = train_f[config.TARGET]
```

```
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```
# Train model
cat_idx = [i for i, c in enumerate(X.columns) if c in config.CAT_COLS]
cb = CatBoostClassifier(iterations=100, cat_features=cat_idx, verbose=50)
cb.fit(X_train, y_train, eval_set=(X_val, y_val))
```
