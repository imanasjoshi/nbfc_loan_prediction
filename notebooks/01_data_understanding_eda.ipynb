# Data Understanding and EDA

This notebook performs data quality analysis and exploratory data analysis.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```
train = pd.read_csv('../data/raw/Train.csv')
test = pd.read_csv('../data/raw/Test.csv')
train.head()
```

```
train.info()
```

```
train['default'].value_counts(normalize=True).plot(kind='bar', title='Default Rate')
```

```
sns.boxplot(x='default', y='loan_amnt', data=train)
plt.title('Loan Amount by Default Status')
```

```
sns.boxplot(x='default', y='interest_rate', data=train)
plt.title('Interest Rate by Default Status')
```
